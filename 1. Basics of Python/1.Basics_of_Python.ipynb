## Exercise 1: Prime Numbers

**Problem:** Write a Python program that checks whether a given number is prime or not.
A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

**Approach:**
- Numbers less than or equal to 1 are not prime.
- Check divisibility from 2 up to the square root of the number (any factor larger than the square root would have a corresponding factor smaller than it, so checking beyond the square root is redundant).
- If no divisor is found in that range, the number is prime.

```python
import math

def is_prime(number):
    """
    Check whether a given integer is a prime number.

    Parameters:
        number (int): The number to check.

    Returns:
        bool: True if the number is prime, False otherwise.
    """
    # Prime numbers are defined as greater than 1
    if number <= 1:
        return False

    # 2 is the only even prime number
    if number == 2:
        return True

    # Eliminate other even numbers quickly
    if number % 2 == 0:
        return False

    # Only need to check odd divisors up to the square root of the number
    for divisor in range(3, int(math.sqrt(number)) + 1, 2):
        if number % divisor == 0:
            return False

    return True


# --- Program entry point ---
# Take input from the user and display whether it is prime
user_number = int(input("Enter a number to check if it is prime: "))

if is_prime(user_number):
    print(f"{user_number} is a prime number.")
else:
    print(f"{user_number} is not a prime number.")
```

### Quick test (optional)
A small sanity check across a range of numbers to confirm the function behaves correctly, without needing manual input.

```python
# Sanity check: print all prime numbers between 1 and 30
primes_up_to_30 = [n for n in range(1, 31) if is_prime(n)]
print("Primes up to 30:", primes_up_to_30)
```

## Exercise 2: Product of Random Numbers

**Problem:** Generate two random numbers, ask the user to enter their product, and check whether the user's answer is correct.

**Approach:**
- Use the `random` module to generate two random integers within a chosen range.
- Compute the actual product.
- Prompt the user for their answer and compare it with the actual product.

```python
import random

def product_quiz(low=1, high=20):
    """
    Generate two random numbers in the range [low, high], ask the user
    for their product, and tell them whether they answered correctly.

    Parameters:
        low (int): Lower bound (inclusive) for the random numbers.
        high (int): Upper bound (inclusive) for the random numbers.
    """
    # Generate two random numbers within the given range
    num1 = random.randint(low, high)
    num2 = random.randint(low, high)

    # Compute the correct product
    correct_product = num1 * num2

    # Ask the user to calculate the product
    user_answer = int(input(f"What is the product of {num1} and {num2}? "))

    # Compare the user's answer with the correct product
    if user_answer == correct_product:
        print("Correct! Well done.")
    else:
        print(f"Incorrect. The correct answer was {correct_product}.")


# --- Program entry point ---
product_quiz()
```

## Exercise 3: Squares of Even/Odd Numbers

**Problem:** Print the squares of all even or odd numbers within the range of 100 to 200.

**Choice documented in code:** This solution prints the squares of **even** numbers between 100 and 200 (inclusive). To switch to odd numbers, change the condition from `number % 2 == 0` to `number % 2 != 0`.

```python
# Choice: printing squares of EVEN numbers between 100 and 200 (inclusive).
# To print ODD numbers instead, change the condition below to `number % 2 != 0`.

print("Squares of even numbers from 100 to 200:")

for number in range(100, 201):  # 201 so that 200 is included
    if number % 2 == 0:  # keep only even numbers
        print(f"{number}^2 = {number ** 2}")
```

## Exercise 4: Hospital Billing System

**Problem Statement:** A hospital charges patients based on:

- Room charges per day: ₹2,000
- Doctor consultation fee (flat): ₹1,500
- Lab test charges: ₹300 per test
- Medicine charges: total cost of medicines provided
- Discount: 10% discount if the total bill (before discount) exceeds ₹10,000

Calculate the final bill for a patient who:
- Stayed for **4 days**
- Had **3 lab tests**
- Had medicine charges of **₹2,400**

**Approach:**
- Define constant per-unit rates.
- Compute each charge component separately, then sum them for the bill before discount.
- Apply the 10% discount only if the pre-discount total exceeds ₹10,000.
- Display an itemized bill along with the final amount.

```python
def calculate_hospital_bill(days_stayed, num_lab_tests, medicine_charges):
    """
    Calculate a hospital patient's final bill.

    Parameters:
        days_stayed (int): Number of days the patient stayed.
        num_lab_tests (int): Number of lab tests taken.
        medicine_charges (float): Total cost of medicines provided.

    Returns:
        dict: A breakdown of charges and the final bill amount.
    """
    # --- Fixed rates ---
    ROOM_CHARGE_PER_DAY = 2000
    CONSULTATION_FEE = 1500       # flat fee, independent of stay length
    LAB_TEST_CHARGE = 300         # per test
    DISCOUNT_THRESHOLD = 10000    # bill must exceed this to qualify for discount
    DISCOUNT_RATE = 0.10          # 10% discount

    # --- Compute individual charge components ---
    room_charges = ROOM_CHARGE_PER_DAY * days_stayed
    lab_charges = LAB_TEST_CHARGE * num_lab_tests

    # --- Total bill before discount ---
    total_before_discount = (
        room_charges + CONSULTATION_FEE + lab_charges + medicine_charges
    )

    # --- Apply discount if applicable ---
    if total_before_discount > DISCOUNT_THRESHOLD:
        discount_amount = total_before_discount * DISCOUNT_RATE
    else:
        discount_amount = 0

    final_bill = total_before_discount - discount_amount

    # Return a detailed breakdown so it can be displayed clearly
    return {
        "Room Charges": room_charges,
        "Consultation Fee": CONSULTATION_FEE,
        "Lab Test Charges": lab_charges,
        "Medicine Charges": medicine_charges,
        "Total Before Discount": total_before_discount,
        "Discount Applied": discount_amount,
        "Final Bill": final_bill,
    }


# --- Program entry point ---
# Patient details as given in the problem statement
days_stayed = 4
num_lab_tests = 3
medicine_charges = 2400

bill = calculate_hospital_bill(days_stayed, num_lab_tests, medicine_charges)

print("----- Hospital Bill Breakdown -----")
for item, amount in bill.items():
    print(f"{item}: \u20b9{amount:,.2f}")
```
