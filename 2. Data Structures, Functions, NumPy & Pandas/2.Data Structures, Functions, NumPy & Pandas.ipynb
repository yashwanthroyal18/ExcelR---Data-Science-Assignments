## Q1. List Operations

Create a list of 5 integers and demonstrate: `append`, `extend`, `insert`, `remove`, `pop`, `clear`, `index`, `count`, `sort`, and `reverse`.

Each operation's effect is printed immediately after it runs so the change is visible step by step.

```python
# Start with a list of 5 integers
numbers = [10, 20, 30, 40, 50]
print("Original list:", numbers)

# append(): adds a single element to the end of the list
numbers.append(60)
print("After append(60):", numbers)

# extend(): adds all elements of another iterable to the end of the list
numbers.extend([70, 80])
print("After extend([70, 80]):", numbers)

# insert(): inserts an element at a specific index
numbers.insert(2, 25)  # insert 25 at index 2
print("After insert(2, 25):", numbers)

# remove(): removes the first matching value from the list
numbers.remove(25)
print("After remove(25):", numbers)

# pop(): removes and returns the element at the given index (last element if no index given)
popped_value = numbers.pop()
print(f"After pop() -> removed {popped_value}:", numbers)

# index(): returns the index of the first occurrence of a value
position = numbers.index(30)
print(f"index(30) -> found at position {position}")

# count(): counts how many times a value appears in the list
numbers.append(30)  # add a duplicate so count() has something meaningful to show
print("List after adding a duplicate 30:", numbers)
occurrences = numbers.count(30)
print(f"count(30) -> {occurrences} occurrence(s)")

# sort(): sorts the list in place, in ascending order by default
numbers.sort()
print("After sort():", numbers)

# reverse(): reverses the order of elements in place
numbers.reverse()
print("After reverse():", numbers)

# clear(): removes all elements from the list
numbers.clear()
print("After clear():", numbers)
```

## Q2. Tuples Are Immutable

Create a tuple of 3 student names and try to change the second name.

**Explanation:** Tuples are *immutable* — once created, their elements cannot be reassigned. Attempting `tuple[index] = value` raises a `TypeError`, unlike a list where item assignment is allowed. This immutability is why tuples are used for fixed collections of data that shouldn't change (e.g., coordinates, records).

```python
# A tuple of 3 student names
students = ("Alice", "Bob", "Charlie")
print("Original tuple:", students)

# Try changing the second name (index 1) and catch the resulting error
try:
    students[1] = "Bobby"
except TypeError as error:
    print(f"Error occurred: {error}")
    print("Explanation: tuples are immutable, so individual elements cannot be reassigned "
          "after the tuple is created.")
```

## Q3. Sets Remove Duplicates

Create a set of integers that includes duplicate values and print it.

**Explanation:** A `set` in Python only stores **unique** elements. When duplicate values are supplied, Python automatically discards the repeats, and the set also does not preserve insertion order (elements are stored based on their hash for fast membership testing).

```python
# Create a set with intentional duplicate values
numbers_set = {1, 2, 2, 3, 4, 4, 4, 5}
print("Set (duplicates removed automatically):", numbers_set)

print("Explanation: even though the value 2 was written twice and 4 was written three "
      "times, the set keeps only one copy of each because sets only store unique elements.")
```

## Q11. Building and Cleaning a DataFrame

Create a small DataFrame manually with 10 rows and columns `'Name'`, `'Age'`, `'City'`, and `'Salary'`.
A couple of missing `Salary` values and one duplicate row are included on purpose so that the
cleaning steps below (filling missing values, dropping duplicates) have something real to act on.

Then:
- Use `.info()` and `.describe()`
- Select the `'Name'` and `'City'` columns
- Drop the `'City'` column
- Fill missing values in `'Salary'` with the column mean
- Remove duplicate rows

```python
# Manually created data: 10 rows, with one duplicate row (rows for "Ivan")
# and a couple of missing Salary values (represented with np.nan) to make the
# cleaning steps meaningful.
data = {
    "Name":   ["Aditi", "Rahul", "Meera", "Karan", "Sara",
               "Ivan",  "Ivan",  "Neha",  "Vikram", "Farah"],
    "Age":    [23, 35, 29, 41, 27, 33, 33, 24, 38, 30],
    "City":   ["Hyderabad", "Delhi", "Mumbai", "Bengaluru", "Chennai",
               "Pune", "Pune", "Kolkata", "Ahmedabad", "Jaipur"],
    "Salary": [45000, 62000, np.nan, 78000, 51000,
               58000, 58000, np.nan, 71000, 49000],
}

df = pd.DataFrame(data)
print("Full DataFrame:")
df
```

```python
# .info() gives column dtypes, non-null counts, and memory usage
print("DataFrame info:")
df.info()
```

```python
# .describe() gives summary statistics for the numeric columns
print("Summary statistics:")
df.describe()
```

```python
# Select only the 'Name' and 'City' columns
name_city = df[["Name", "City"]]
print("Name and City columns:")
name_city
```

```python
# Drop the 'City' column (axis=1 means we're dropping a column, not a row)
df_no_city = df.drop(columns=["City"])
print("DataFrame after dropping 'City':")
df_no_city
```

```python
# Fill missing values in 'Salary' with the column's mean
salary_mean = df["Salary"].mean()
print(f"Mean salary (ignoring missing values): {salary_mean:.2f}")

df["Salary"] = df["Salary"].fillna(salary_mean)
print("\nDataFrame after filling missing Salary values:")
df
```

```python
# Remove duplicate rows (the two identical "Ivan" rows collapse into one)
print(f"Row count before removing duplicates: {len(df)}")

df_deduped = df.drop_duplicates()

print(f"Row count after removing duplicates: {len(df_deduped)}")
print("\nFinal cleaned DataFrame:")
df_deduped
```
